## 4. Tokenize words Using NLTK

```python
# word tokenizer -> the most used method in nltk
import nltk.tokenize as tkn
def nltk_tokenizer(t) -> list:
    words=tkn.word_tokenize(t)
    return words

nltk_tokenizer(text)
```

```python
# word punctuation tokenizer 
def nltk_punc_tokenizer(t) -> list:
    words=tkn.wordpunct_tokenize(t)
    return words

nltk_punc_tokenizer(text)
```

```python
# TokTok tokenizer 
def nltk_TokTok_tokenizer(t) -> list:
    words=tkn.ToktokTokenizer().tokenize(t)
    return words

nltk_TokTok_tokenizer(text)
```

```python
# tree bank word tokenizer
def tree_bank_word_tokenizer(t) -> list:
    words=tkn.TreebankWordTokenizer().tokenize(t)
    return words

tree_bank_word_tokenizer(text)
```

## 2. Stemming words using Regular Expression

```python
from nltk.stem import RegexpStemmer
def reg_stemmer(t) -> list:
    reg_stem=RegexpStemmer('ing$|ed$|s$|able$|ive$|en$|tion$', min=4)
    return [reg_stem.stem(word) for word in t]

data=remove_stop_words_punc(regex_find_tokenize(text))
reg_stemmer(data)
```

## 3. Stemming words using Porter Stemmer

This is the Porter stemming algorithm. It follows the algorithm presented in

Porter, M. “An algorithm for suffix stripping.” Program 14.3 (1980): 130-137.

with some optional deviations that can be turned on or off with the mode argument to the constructor.

```python
from nltk.stem import PorterStemmer
def porter_stem(t) -> list:
    stem_porter=PorterStemmer()
    return [stem_porter.stem(word) for word in t]

data=remove_stop_words_punc(regex_find_tokenize(text))
porter_stem(data)
```

## 4. Stemming words using Lancaster Stemmer

A word stemmer based on the Lancaster (Paice/Husk) stemming algorithm. Paice, Chris D. “Another Stemmer.” ACM SIGIR Forum 24.3 (1990): 56-61.

```python
from nltk.stem import LancasterStemmer
def lancaster_stem(t) -> list:
    lancaster_stem=LancasterStemmer()
    return [lancaster_stem.stem(word) for word in t]
data=remove_stop_words_punc(regex_find_tokenize(text))
lancaster_stem(data)
```

## 5. Stemming using SnowballStemmer

```python
from nltk.stem import snowball
def snowball_stem(t) -> list:
    snow_stem=snowball.EnglishStemmer()
    return [snow_stem.stem(word) for word in t]

data=remove_stop_words_punc(regex_find_tokenize(text))
snowball_stem(data)
```

## 6. Explanation of the difference between All taken algorithms

* regular expression method stems the words based on a given suffix, generally considered to be susciptable to accuracy and quality errors
* Porter Stemmer is the most used algorithm for stemming the words, considered to be more accurate than other algorithms
* Lancaster is more aggresive than porter stemmer, resulting in less accurate, sentiment corrupted words
* Lancaster is faster than Porter algorithm, yet porter is widely used as it is more accurate
* Snowball is faster and more aggresive than Porter and less aggreive than lancaster, usually considered as the optimal algorithm to be deployed
* Snowball supports different wide range of languages making it is the ideal choice for many scenarios
