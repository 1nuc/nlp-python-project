## 4. Tokenize words Using NLTK

```python
# word tokenizer -> the most used method in nltk
import nltk.tokenize as tkn
def nltk_tokenizer(t) -> list:
    words=tkn.word_tokenize(t)
    return words
```

```python
# word punctuation tokenizer 
def nltk_punc_tokenizer(t) -> list:
    words=tkn.wordpunct_tokenize(t)
    return words
```

## 7. Stop words and panctuation Remmoval

```python
# download stop words list from nltk library
import nltk
import string
nltk.download('stopwords')
```

```python
def remove_stop_words_punc(t) -> list:
    stop_words=nltk.corpus.stopwords.words('english')
    punc=string.punctuation
    return [word.lower() for word in t if word.lower() not in stop_words and word not in punc]
```
