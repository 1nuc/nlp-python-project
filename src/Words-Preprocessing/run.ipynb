## 8. Stop words and panctuation Remmoval

```python
# download stop words list from nltk library
import nltk
import string
nltk.download('stopwords')
```

```python
def remove_stop_words_punc(t) -> list:
    stop_words=nltk.corpus.stopwords.words('english')
    punc=string.punctuation
    return [word.lower() for word in t if word.lower() not in stop_words and word not in punc]

remove_stop_words_punc(regex_find_tokenize(data_1))
```

## 9. The importance of filtering stop word and punctuation in text analytics

#### Why Filter Stop Words in text analytics 
- Enhancing Model Performance
    * Stopwords can introduce noise into the analysis
    * Removing them allows the model to focus on meaningful words, improving accuracy

- Reducing Dimensionality
    * Stopwords add unnecessary tokens, increasing the dataset size
    * By removing stopwrods, the dimensionality of the feature space is reduced, making models more efficient

#### Why filter punctuation in text analytics
- Reduces Noise: Punctuation often adds unnecessary complexity to text analysis
- Enhances Tokenizaiton: Simplifies splitting and processinf of text

#### When punctuation is important 
- Sentiment Analysis: Emojis and exclamaition marks can indicate sentiment
- Named Entity Recognition: Hyphenated words (e.g., "state-of-the-art") may need preservation

## 2. Stemming words using Regular Expression

```python
from nltk.stem import RegexpStemmer
def reg_stemmer(t) -> list:
    reg_stem=RegexpStemmer('ing$|ed$|s$|able$|ive$|en$|tion$', min=4)
    return [reg_stem.stem(word) for word in t]

data=remove_stop_words_punc(regex_find_tokenize(data_1))
reg_stemmer(data)
```

## 3. Stemming words using Porter Stemmer

This is the Porter stemming algorithm. It follows the algorithm presented in

Porter, M. “An algorithm for suffix stripping.” Program 14.3 (1980): 130-137.

with some optional deviations that can be turned on or off with the mode argument to the constructor.

```python
from nltk.stem import PorterStemmer
def porter_stem(t) -> list:
    stem_porter=PorterStemmer()
    return [stem_porter.stem(word) for word in t]

data=remove_stop_words_punc(regex_find_tokenize(data_1))
porter_stem(data)
```

## 4. Stemming words using Lancaster Stemmer

A word stemmer based on the Lancaster (Paice/Husk) stemming algorithm. Paice, Chris D. “Another Stemmer.” ACM SIGIR Forum 24.3 (1990): 56-61.

```python
from nltk.stem import LancasterStemmer
def lancaster_stem(t) -> list:
    lancaster_stem=LancasterStemmer()
    return [lancaster_stem.stem(word) for word in t]
data=remove_stop_words_punc(regex_find_tokenize(data_1))
lancaster_stem(data)
```

## 5. Stemming using SnowballStemmer

```python
from nltk.stem import snowball
def snowball_stem(t) -> list:
    snow_stem=snowball.EnglishStemmer()
    return [snow_stem.stem(word) for word in t]

data=remove_stop_words_punc(regex_find_tokenize(data_1))
snowball_stem(data)
```

## 6. Explanation of the difference between All taken algorithms

* regular expression method stems the words based on a given suffix, generally considered to be susciptable to accuracy and quality errors
* Porter Stemmer is the most used algorithm for stemming the words, considered to be more accurate than other algorithms
* Lancaster is more aggresive than porter stemmer, resulting in less accurate, sentiment corrupted words
* Lancaster is faster than Porter algorithm, yet porter is widely used as it is more accurate
* Snowball is faster and more aggresive than Porter and less aggreive than lancaster, usually considered as the optimal algorithm to be deployed
* Snowball supports different wide range of languages making it is the ideal choice for many scenarios

## 4. Regular Expression tagger

```python
import re
from nltk.tag import RegexpTagger
def reg_tag(t):    
    pattern=[
     (r'/d+$', 'CD'), # cardinal numbers
     (r'(The|a|the|A|An|an)$', 'DT'), # determines
     (r'.*(ble|ive|ant|ous|y|cal|ic|ent|ish|ate|ome|ful)$', 'JJ'), # adjectives
     (r'.*(ly|ally)$','RB'), #adverbs
     (r'.*ed','VBD'), # past tense verb
     (r'.*ing', 'VBG'), # gerunds
     (r'(and|but|or)$','CC'),
     (r'(at|down|up|between|around|by|with|in|on|of{1,2}|under|across|along|around|above|against|beside|about|after|before|behind|next)', 'IN'),
     (r'.*(ness|less|ment|age|ity)?$', 'NN') # nouns
    ]
    reg_tag=RegexpTagger(pattern)
    return reg_tag.tag(t)

reg_tag(regex_find_tokenize(pos_data))
```

## 2. Manual computation of the sentence probability using unsmoothed bigram model

### Unsmooth bigram model means including zeros in the calculation.
### Sentence: " I read a different book by Danielle "
#### Step by step calculation:
* (I, <start>) => (<start>, I) /<start> => 1/3 = 0.33
* (read, I) => (I, read)/(I) => 1/1 = 1
* (a, read) => (read,a)/(read) => 3/3 = 1
* (different, a) => (a, different)/(a) => 1/3 = 0.33
* (book, different)=>(different, book) /(different) => 1/1 = 1
* (by, book)=>(book,by) /(book) => 1/3 = 0.33
* (Danielle, by) => (by, Danielle) / (by) => 1/1 = 1
* (<end>, Danielle) => (Danielle, <end>) / (Danielle) => 1/1 =1

<h4> Sentence probability = 0.33 * 1 * 1 * 0.33 * 1 * 0.33 * 1 * 1 = 0.035937

## 3. Manual computation of the sentence probability using smoothed bigram model

### smooth bigram model means adding the word count in the probability of the word occurence divided by the probability of the dominator plus the total count of possible next words.
#### Begining of a word is not considered as the first word cannot be determined
### Sentence: " I read a different book by Danielle "
#### Step by step calculation:
* (I, <start>) => (<start>, I)+1 /<start>+8 => 2/11 = 0.182
* (read, I) => (I, read) + 1/(I) +8 => 2/9 = 0.222
* (a, read) => (read,a)+1/(read) +8 => 4/11 = 0.364

## bigram modelling

```python
import numpy as np
from functools import reduce

def unsmooth_bgram(x, bg):
    probs=[]
    for gram in bg:
        train_text=x
        total_count=x.count(gram[0])
        pair=' '.join(tuple((gram)))
        word_pair_count=' '.join(train_text).count(pair)
        probs.append(word_pair_count /total_count)
    return np.prod(probs)
```

```python
import numpy as np
from functools import reduce

def smooth_bgram(x, bg):
    probs=[]
    for gram in bg:
        train_text=x
        total_count=x.count(gram[0]) + len(bg)
        pair=' '.join(tuple((gram)))
        word_pair_count=' '.join(train_text).count(pair) +1
        probs.append(word_pair_count /total_count)
    return np.prod(probs)
```
