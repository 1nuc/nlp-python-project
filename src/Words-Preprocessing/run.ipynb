## 4. Tokenize words Using NLTK

```python
# word tokenizer -> the most used method in nltk
import nltk.tokenize as tkn
def nltk_tokenizer(t) -> list:
    words=tkn.word_tokenize(t)
    return words

nltk_tokenizer(text)
```

```python
# word punctuation tokenizer 
def nltk_punc_tokenizer(t) -> list:
    words=tkn.wordpunct_tokenize(t)
    return words

nltk_punc_tokenizer(text)
```

```python
# TokTok tokenizer 
def nltk_TokTok_tokenizer(t) -> list:
    words=tkn.ToktokTokenizer().tokenize(text)
    return words

nltk_TokTok_tokenizer(text)
```

```python
# tree bank word tokenizer
def tree_bank_word_tokenizer(t) -> list:
    words=tkn.TreebankWordTokenizer().tokenize(t)
    return words

tree_bank_word_tokenizer(text)
```

## 7. Stop words and panctuation Remmoval

```python
# download stop words list from nltk library
import nltk
import string
nltk.download('stopwords')
```

```python
def remove_stop_words_punc(t) -> list:
    stop_words=nltk.corpus.stopwords.words('english')
    punc=string.punctuation
    return [word.lower() for word in t if word.lower() not in stop_words and word not in punc]

remove_stop_words_punc(regex_find_tokenize(text))
```

## 2. Stemming words using Regular Expression

```python
from nltk.stem import RegexpStemmer
def reg_stemmer(t) -> list:
    reg_stem=RegexpStemmer('ing$|ed$|s$|able$|ive$|en$|tion$', min=4)
    return [reg_stem.stem(word) for word in t]

data=remove_stop_words_punc(regex_find_tokenize(text))
reg_stemmer(data)
```
